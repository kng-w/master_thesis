```python
import matplotlib.pyplot as plt
from matplotlib import rcParams

# フォントの設定をArialに変更
rcParams["font.family"] = "sans-serif"

# 表のデータ
categories = ["A", "B", "C", "D", "E"]
chatgpt_scores = [3, 1, 2, 1, 1]
ai_scores = [2, 3, 5, 4, 4]

# 散布図の作成
plt.scatter(chatgpt_scores, ai_scores)

# 各点にカテゴリ名をラベルとして追加
# 重なっている点を少しずらすためのオフセット
offsets = [(0.1, 0.1), (-0.2, -0.2), (0.1, -0.2), (-0.2, 0.2), (0.1, 0.2)]
for i, category in enumerate(categories):
    plt.text(
        chatgpt_scores[i] + offsets[i][0],
        ai_scores[i] + offsets[i][1],
        category,
        fontsize=9,
    )

# 軸のラベルを追加
plt.xlabel("ChatGPTの使用頻度")
plt.ylabel("AIへの期待度")

# グラフのタイトル

# 軸の範囲を設定（余白を持たせる）
plt.xlim(0.5, 5.5)
plt.ylim(0.5, 5.5)

# 目盛りのラベルのみを表示し、目盛り線は非表示にする
plt.tick_params(
    axis="both",  # 両軸に適用
    which="both",  # 両方の目盛りに適用（メジャーとマイナー）
    bottom=True,  # 下の目盛りを表示
    top=False,  # 上の目盛りを非表示
    left=True,  # 左の目盛りを表示
    right=False,  # 右の目盛りを非表示
    labelbottom=True,  # 下のラベルを表示
    labelleft=True,  # 左のラベルを表示
    tick1On=False,  # 軸に沿った目盛り線を非表示
    tick2On=False,  # 軸の反対側の目盛り線を非表示
)

# グリッドを非表示
plt.grid(False)

# グラフを表示
plt.show()
```
